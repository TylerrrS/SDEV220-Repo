```python
#Tyler Smekens
#SDEV 220, 11-9-26
#File Name: Student GPA Check
#Description: application will accept a last and first name, then a float as a GPA.
#If the GPA given is 3.25 or more, the app will inform the user that the student has made Honor Roll.
#If the GPA's 3.5 or higher, the app tells them that they've made the Dean's List.

#Create while loop which asks for first & last name + GPA
while True:
    #Get last name. Quit program if "ZZZ" is entered
    lastName = input("Welcome! Please enter student's last name [ZZZ to quit]: ")
    if lastName == "ZZZ":
        break

    #Get first name
    firstName = input("Now, please enter the student's first name: ")

    #Get GPA and convert to float
    gpaStr = input("Enter the student's GPA: ")
    GPA = float(gpaStr)

    #Print a message if student meets Honor Roll requirements
    if GPA >= float(3.25) and GPA < float(3.5):
        print("This student has made the Honor Roll. Great job!")

    #print message if student meets Dean's List reqs
    if GPA >= float(3.5):
        print("Congratulations! This student has made the Dean's List!")
```
