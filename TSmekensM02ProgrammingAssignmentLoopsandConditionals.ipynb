```python
#Tyler Smekens, Module 2 Programming Assignment - Loops and Conditionals, SDEV220, 11-9-26

#4.1 Choose a number between 1 and 10 and assign it to the variable secret. 
#Then, select another number between 1 and 10 and assign it to the variable guess. 
#Next, write the conditional tests (if, else, and elif) to print the string 'too low' if guess 
#is less than secret, 'too high' if greater than secret, and 'just right' if equal to secret.

#Assign number to "secret" & "guess" variables
secret = 8
guess = 5
#Create if/else to determine if secret == guess
if guess == secret:  #if guess is the same number as secret
    print("just right")
elif guess < secret: #if guess is a smaller number than secret
    print("too low")
else:                #if guess is a bigger number than secret
    print("too high")
```

```python
#6.2 Assign the value 7 to the variable guess_me, and the value 1 to the variable number. 
#Write a while loop that compares number with guess_me. Print 'too low' if number is 
#less than guess me. If number equals guess_me, print 'found it!' and then exit the loop. 
#If number is greater than guess_me, print 'oops' and then exit the loop. 
#Increment number at the end of the loop.

#assign "guess_me" and "number" variables  
guess_me = 7
number = 1
#Create while loop comparing varNum to guess_me
#Loop continues as long as number doesn't = guess_me
while number != guess_me:
    if number < guess_me:
        print("Too low!")      #Indicates guessed number is too low, increments by 1
        number += 1
    if number == guess_me:     #Prints once guessed number matches targeted number
        print("Found it!")
        break
    if number > guess_me:      #Prints if guessed number is somehow incremented to be bigger than target number
        print("Oops!")
        break
    
```

```python
#6.3 Assign the value 5 to the variable guess_me. 
#Use a for loop to iterate a variable called number over range(10). 
#If number is less than guess_me, print 'too low'. 
#If it equals guess_me, print found it! and then break out of the for loop. 
#If number is greater than guess_me, print 'oops' and then exit the loop.

#assign "guess_me" and "number" variables
guess_me = 5
number = 0
#Create for loop, testing number in a range of 0-10
for number in range(10):
    if number == guess_me: #Prints when guessed number is reached, then breaks loop
        print("Found it!")
        break
    if number < guess_me:  #Prints if guessed number is too low, continues loop
        print("Too low!")
    if number > guess_me:  #Prints if guessed number somehow goes higher, breaks loop
        print("Oops!")
        break
```
